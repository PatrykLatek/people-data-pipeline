# 01 - Bronze Ingestion

Loads raw employee data from the source CSV file
and stores it as a Delta table in the Bronze layer.

## Setup

```python
%sql
USE CATALOG workspace;
USE SCHEMA people_data_bronze;
```

## Source Path

```python
source_path = "/Volumes/workspace/people_data_bronze/source_files/employees.csv"
```

## Read Source CSV

```python

employees_raw_df = spark.read.csv(
    source_path,
    header=True,
    inferSchema=False
)
display(employees_raw_df)
```

## Basic Validation

```python
employees_raw_df.printSchema()
```

```python
print("Liczba rekordów:", employees_raw_df.count())
```

## Write Bronze Delta Table

```python
(
    employees_raw_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("employees_raw")
)
```

## Final Validation

```python
%sql
select *
from employees_raw
```

```python
%sql
SELECT count(*) as total_records
FROM employees_raw
```

## Summary Tables

```python
%sql
CREATE OR REPLACE TABLE department_summary AS
SELECT
    department_id,
    count(*) AS employee_count,
    round(avg(salary),2) AS  average_salary
FROM people_data_silver.employees_clean
GROUP BY department_id
```

```python
%sql
CREATE OR REPLACE TABLE country_summary AS
SELECT
    country,
    count(*) as employee_count,
    round(avg(salary),2) as average_salary
FROM people_data_silver.employees_clean
GROUP BY country
```

```python
%sql
CREATE OR REPLACE TABLE employment_type_summary AS
SELECT
    employment_type,
    count(*) AS employee_count,
    round(avg(salary),2) AS average_salary
FROM
    people_data_silver.employees_clean
GROUP BY
    employment_type
```

```python
%sql
CREATE OR REPLACE TABLE status_summary AS
SELECT
    status,
    count(*) AS employee_count,
    round(count(*) * 100.0 / (select count(*) from people_data_silver.employees_clean),2) AS percentage
FROM people_data_silver.employees_clean
GROUP BY status
```

```python
%sql
CREATE OR REPLACE TABLE hiring_summary AS
SELECT
    year(hire_date) AS hire_year,
    count(*) AS employee_count
FROM people_data_silver.employees_clean
GROUP BY year(hire_date)
```
